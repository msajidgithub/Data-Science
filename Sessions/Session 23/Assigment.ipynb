```python
from asyncio import QueueEmpty
import sqlite3 as sql
import pandas as pd
import numpy as np

conn = sql.connect('database.sqlite')
conn

query = ""
def sql_query_fun(query):
    return pd.read_sql_query(query, conn)

sql_query_fun("select * from sqlite_master")
```

```python
# Job titles with more than 5 employees and average TotalPay greater than 100k
query = "SELECT JobTitle, COUNT(*) AS EmployeeCount, AVG(TotalPay) AS Average_TotalPay FROM salaries GROUP BY JobTitle HAVING COUNT(*) > 5 AND AVG(TotalPay) > 100000"
sql_query_fun(query)
```

```python
# Employees with average BasePay greater than 50k by JobTitle

query =  "SELECT Year, COUNT(*) AS Employee_Count FROM salaries GROUP BY Year HAVING COUNT(*) > 10" 
sql_query_fun(query)
```

```python
# Employees with average BasePay greater than 50k by JobTitle

query = "SELECT JobTitle, AVG(BasePay) AS 'Average BasePay' FROM salaries GROUP BY JobTitle HAVING AVG(BasePay) > 50000"
sql_query_fun(query)
```
