```python
example_set = {1,2,3,4,5,6,7}

dictionary = {}

for value in example_set:
    if value % 2 == 0:
        dictionary[value] = value*2
    else:
        dictionary[value] = value*3
print(dictionary)
```

```python
def get_avg(numbers):
    get_sum = sum(numbers)
    get_len = len(numbers)
    return  get_sum / get_len


print(get_avg([10, 20, 30, 40, 50]))
```

```python

def get_avg_new(arr):
    total = 0
    
    for ar in arr:
        total = total + ar 

    return total / len(arr)

print(get_avg_new([10, 20, 30]))
```

```python
def get_even_sum(newArr):
    total = 0

    for i in newArr:

        if i % 2 == 0:
            total = total + i

    return total

print(get_even_sum([1, 2, 3, 4, 5, 6]))
```
