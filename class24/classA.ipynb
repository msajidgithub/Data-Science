```python
import sqlite3 as sql
import pandas as pd

conn = sql.connect('factbook.db')
conn

query = ""
def sql_query_fun(query):
    return pd.read_sql_query(query, conn)
    
query = "SELECT * FROM sqlite_master"
sql_query_fun(query)
```
